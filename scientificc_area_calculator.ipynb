```python
#scientific area calculator
print("Enter 1 for square, 2 for rectangle, 3 for traingle,4 for parallelogram,5 for rhombus, 6 for trapezium,7 for circle,8 for semicircle,9 for elipse,10 for equilateral triangle,11 for kite, 12 for hexagone,13 for heptagone,14 for octagon")
choice=int(input("Enter your choice number"))
if choice==1:
  side=(int(input("Enter the side of sqaure")))
  area=side*side
  print("The area of square is",area)
elif choice==2:
  length=int(input("Enter the lenght"))
  breadth=int(input("Enter the breadth"))
  area=length*breadth
  print("The area of rectangle is",area)
elif choice==3:
   length=int(input("Enter the lenght"))
   base=int(input("Enter the base"))
   area=1/2*base*length
   print("The area of triangle is",area)
elif choice==4:
  base=int(input("Enter the base"))
  height=int(input("Enter the height"))
  area=height*base
elif choice==5:
  diagonal1 =int(input("Enter diagonal1"))
  diagonal2=int(input("Enter diagonal2"))
  area=1/2*diagonal1*diagonal2
  print("The area of rhombus is",area)
elif choice==6:
  lenght1=int(input("Enter the lenght1"))
  lenght2=int(input("Enter the length2"))
  height=int(input("Enter the height"))
  area=1/2*height*(lenght1+lenght2)
  print("The area of trapezium is",area)
elif choice==7:
  import math
  radius=int(input("Enter the radius"))
  area= math.pi*(radius*radius)
  print("The area of circle is",area)
elif choice==8:
  import math
  radius =int(input("Enter the radius"))
  area=math.pi*radius*radius/2
  print("The area of semicircle is",area)
elif choice==9:
  import math
  major_axis=int(input("Enter the major axis"))
  minor_axis=int(input("Enter the minor axis"))
  area=math.pi*major_axis*minor_axis
  print("The area of elipse is",area)
elif choice==10:
  import math
  side=int(input("Enter the side"))
  area=(math.sqrt(3)/4)*side*side
  print("The area of equilateral traingle is",area)
elif choice==11:
  short_diagonal=int(input("Enter the short diagonal"))
  long_diagonal=int(input("Enter the long diagonal"))
  area=1/2*short_diagonal*long_diagonal
  print("The area for kite is",area)
elif choice==12:
  import math
  side=int(input("Enter the side"))
  area=3*math.sqrt(3)/2*side*side
  print("The area of hexagon is",area)
elif choice==13:
  import math
  side=int(input("Enter the side"))
  area=(7/4)*side*side*math.tan(7/180)
  print("The area of heptagon is",area)
elif choice==14:
  import math
  side=int(input("Enter the side"))
  area=2*side*side*(1/math.sqrt(2))
  print("The area of octagon is",area)








```
